### For Loop

```python
# ১. ১ থেকে ১০ পর্যন্ত প্রিন্ট করো।

for i in range(1, 11):
    print(i)
```

```python
# ২. ১ থেকে ১০০ পর্যন্ত সব জোড় সংখ্যা প্রিন্ট করো।

for i in range(0, 100, 2):
    print(i)
```

```python
# ৩. ১ থেকে ৫০ পর্যন্ত সব বেজোড় সংখ্যার যোগফল বের করো।

for i in range(1, 52,2):
    print(i)
```

```python
# ৪. User থেকে একটি সংখ্যা নাও। ১ থেকে সেই সংখ্যা পর্যন্ত multiplication table প্রিন্ট করো।

num = int(input('Display Multiplication Table'))
for i in range(1, 11):
    print(f'{num} x {i} = {num * i}')
```

```python
# ৫. একটি string নাও। প্রতিটি character আলাদা লাইনে প্রিন্ট করো।

for char in 'Python':
    print(char)
```

```python

# ১০. ১ থেকে N পর্যন্ত সংখ্যার মধ্যে কতটা জোড় আর কতটা বেজোড় — count করো।

n = int(input('Enter a N number: '))

even_count = 0
odd_count = 0

for i in range(1, n + 1):
    if i % 2 == 0:
        even_count += 1

    else:
        odd_count += 1

print('Even', even_count)
print('Odd', odd_count)
```
