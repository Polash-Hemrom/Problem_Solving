### 🟢 Variables & Data Types

```python
# ১. একটি ভেরিয়েবলে তোমার নাম, বয়স, এবং GPA স্টোর করো। প্রতিটির type() প্রিন্ট করো।

name = 'Polash'
age = 30

print(type(name))
print(type(age))
```

```python
# ২. একটি integer কে float এ, তারপর string এ কনভার্ট করো। প্রতিটি স্টেপে value ও type প্রিন্ট করো।

number = 25
print(type(number))

integer_to_float = float(number)
print(type(integer_to_float))

float_to_string = str(integer_to_float)
print(type(float_to_string))
```

#### 🟡 Input / Output

```python
# ৪. User এর কাছ থেকে নাম ও শহর নাও, তারপর প্রিন্ট করো:

name = input('Enter your name : ')
city = input('Enter your city : ')

print(f'Hello {name}, you are from {city}')
```

```python
# ৫. দুইটি সংখ্যা input নাও এবং তাদের যোগফল, গুণফল প্রিন্ট করো। (Type conversion লাগবে)

x = int(input('Enter your favourite number: '))
y = int(input('Enter your favourite number: '))

addition = x + y
multiplication = x * y

print(addition)
print(multiplication)
```

```python
# ৬. User এর জন্মসাল নাও, বর্তমান সাল 2026 ধরে বয়স বের করো এবং সুন্দরভাবে প্রিন্ট করো।

present_year = int(input('Enter Present Year: '))
birth_of_date = int(input('Enter Your Birth Date: '))


age_calculate = ( present_year - birth_of_date)
print(age_calculate)
```
