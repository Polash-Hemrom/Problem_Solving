### 🟠 Operators

```python
# ৭. দুইটি সংখ্যা নাও। এদের +, -, *, /, //, %, ** — সবগুলো অপারেটরের result প্রিন্ট করো।

x = 25
y = 10

addition = x + y
print(addition)

subtraction = x - y
print(subtraction)

multiplication = x * y
print(multiplication)

divide = x / y
print(divide)

floor_divide = x // y
print(floor_divide)

modulus = x % y
print(modulus)

power = x ** y
print(power)
```

```python
# ৮. তিনটি সংখ্যা x, y, z নাও। Check করো — x > y এবং y > z দুটোই সত্য কিনা (and অপারেটর দিয়ে)।

x = 30
y = 20
z = 10

check = x > y  and y > z
print(check)
```

```python
# ৯. একটি সংখ্যা নাও। সেটা কি ১০ এর বেশি অথবা ৫ এর কম? or দিয়ে check করো এবং True/False প্রিন্ট করো।

x = 20

checking = x > 10 or x < 5
print(checking)
```

```python
# a = 15 এবং b = 4। এখন a // b এবং a % b বের করো। তারপর verify করো:
# (a // b) * b + (a % b) == a — এটা কি True?

a = 15
b = 4

print(a / b)
print(a // b)
print(a % b)

check = (a // b) * b + (a % b) == a
print(bool(check))
```

```python
# দুইটি সংখ্যা নাও। Comparison অপারেটর দিয়ে check করো — ==, !=, >, <, >=, <= সবগুলোর result প্রিন্ট করো।

a = int(input('Enter your number: '))
b = int(input('Enter your number: '))

equal = a == b
print(equal)

not_equal = a != b
print(not_equal)

greater = a > b
print(greater)

less = a < b
print(less)

greater_than_equal = a >= b
print(greater_than_equal)

less_than_equal = a <= b
print(less_than_equal)
```

```python
# ১২. একটি সংখ্যা নাও। সেটা positive, negative নাকি zero — সেটা প্রিন্ট করো।

x = int(input('Enter your number: '))
if x > 1:
    print('Positive')
elif x < 0:
    print('Negative')

else:
    print('Zero')
```

```python
# ১৩. User একটি সংখ্যা দেবে। সেটা জোড় না বেজোড় — check করো। (% অপারেটর ব্যবহার করো)

x = int(input('Enter a Number: '))

if x % 2 == 0:
    print('Even')
else:
    print('Odd')
```

```python
# ১৪. একটি ক্যালকুলেটর বানাও — দুইটি সংখ্যা ও একটি অপারেটর (+, -, *, /) নাও, result বের করো। Division এ 0 হলে error message দাও।

num = int(input('Enter Your number: '))
num2 = int(input('Ener Your number: '))

operator = input ('+, -, *, /')

if operator == '+':
    print(num + num2)

elif operator == '-':
    print(num - num2)

elif operator == '*':
    print(num * num2)

elif operator == '/':
    if num and num2 % 2 == 0:
        print('Error')
```

```python
# ১৫. ★ Mini Project: User এর marks নাও (০-১০০)। Grade বের করো:

""" ৮০+ → A
৬০-৭৯ → B
৪০-৫৯ → C
৪০ এর নিচে → F
"""

marks = int(input('Enter your marks: '))

if marks >= 80:
    print('A')
elif marks >= 60:
    print('B')
elif marks >= 40:
    print('C')
else:
    print('F')
```

```python
# User একটি বছর দেবে। সেটা leap year কিনা check করো।
# (Leap year: ৪ দিয়ে ভাগ যায়, কিন্তু ১০০ দিয়ে গেলে leap year না — তবে ৪০০ দিয়ে গেলে আবার leap year)

year = int(input('Enter your year: '))

if year % 4 == 0 and year % 100 !=0:
    print('Leap Year')
else:
    print('Not Leap year')
```

```python
# তিনটি সংখ্যা নাও। Check করো তিনটি মিলে triangle হতে পারে কিনা।
# (Rule: যেকোনো দুই বাহুর যোগফল তৃতীয় বাহু থেকে বড় হতে হবে)

a = int(input('Enter number: '))
b = int(input('Enter number: '))
c = int(input('Enter number: '))

if a + b > c and a + c > b and b + c > a:
    print('Triangle Possible')

else:
    print('Not a Triangle')
```

```python
# User এর মাসিক income নাও। Tax calculate করো:

# ৫০,০০০ এর নিচে → Tax নাই
# ৫০,০০০–১,০০,০০০ → ৫% tax
# ১,০০,০০০ এর উপরে → ১০% tax

income = int(input('Enter your income: '))

if income < 50000:
    print('Zero Tax')
elif income >= 50000 and income <= 100000:
    print('5 % Tax')
elif income >= 100000:
    print(' 10% Tax')
```

```python
# User একটি character দেবে। সেটা vowel (a, e, i, o, u) নাকি consonant — check করো। Uppercase/lowercase দুটোই handle করো।

character = input('Enter Text: ')


character.lower()
if character in 'aeiou':
    print('Vowel')
else:
    print('Consonent')
```
